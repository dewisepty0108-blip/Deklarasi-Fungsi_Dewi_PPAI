```python
# Deklarasi fungsi
def perkalian(bil1, bil2):
    # Menghitung perkalian 2 bilangan bulat
    hasil = bil1 * bil2
    return hasil

# Program Utama
bil1 = int(input('Input bil 1 : '))
bil2 = int(input('Input bil 2 : '))

# Memanggil fungsi
print('Hasil Perkalian ', bil1, ' dengan ', bil2, ' adalah ',
      perkalian(bil2=bil1, bil1=bil2))
```
